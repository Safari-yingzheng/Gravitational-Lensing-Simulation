```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from matplotlib.colors import TwoSlopeNorm
import os

plt.rcParams['font.size'] = 12
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['figure.dpi'] = 100

os.makedirs('figures', exist_ok=True)
# 图1：强引力透镜光线偏折路径（EIS 模型）
fig, ax = plt.subplots(figsize=(8, 6))

theta_E = 1.5
e = 0.3
q = 1 - e

x = np.linspace(-10, 10, 1200)

for y0 in [5.5, -5.5]:
    sign = -np.sign(y0)
    transition = 0.5 * (1 + np.tanh(x / 1.2))
    deflection = sign * theta_E * 2.2 * (1 / q) * transition
    y = y0 + deflection
    ax.plot(x, y, 'b-', lw=2, alpha=0.85)

ax.plot(0, 0, 'ro', ms=15, label='Lens Galaxy (EIS)', zorder=5)
ax.plot(-8, 7.5, '*', color='orange', ms=20,
        label='Background Source', zorder=5)
ax.plot(8, 0, 's', color='green', ms=12,
        label='Observer (Earth)', zorder=5)

theta = np.linspace(0, 2 * np.pi, 400)
ax.plot(theta_E * np.cos(theta),
        theta_E * q * np.sin(theta),
        color='red', ls='--', lw=1.2, alpha=0.6,
        label='Einstein ellipse')

ax.axhline(0, color='gray', ls=':', lw=0.8)
ax.axvline(0, color='gray', ls=':', lw=0.8)

ax.set_xlabel('Distance (arbitrary units)')
ax.set_ylabel('Height (arbitrary units)')
ax.set_title('Strong Gravitational Lensing: Light Ray Paths (EIS)',
             fontsize=13, fontweight='bold')
ax.legend(loc='upper right', framealpha=0.9)
ax.grid(True, alpha=0.3)
ax.set_xlim(-10, 10)
ax.set_ylim(-8.5, 9)

plt.tight_layout()
plt.savefig('figures/fig1_light_ray_paths.png', dpi=300, bbox_inches='tight')
plt.show()
```
