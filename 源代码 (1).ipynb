```python
import numpy as np
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 6))
x = np.linspace(-10, 10, 200) 
y_top = 0.15 * x**2 + 1.5 
y_bottom = 0.15 * x**2 - 1.5 
plt.plot(x, y_top, 'b-', linewidth=1.5, label='Light Ray 1')
plt.plot(x, y_bottom, 'b-', linewidth=1.5, label='Light Ray 2')
plt.scatter(0, 0, color='red', s=200, label='Lens Galaxy', zorder=5) 
plt.scatter(-8, 8, color='orange', s=150, marker='*', label='Background Source', zorder=5) 
plt.scatter(8, 0, color='green', s=150, marker='s', label='Observer (Earth)', zorder=5) 
plt.title('Strong Gravitational Lensing: Light Ray Paths Simulation', fontsize=14, fontweight='bold')
plt.xlabel('Distance (arbitrary units)', fontsize=12)
plt.ylabel('Height (arbitrary units)', fontsize=12)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.8)
plt.legend(loc='upper center')
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.savefig('Figure1_RayTracing.png', dpi=300)
plt.show()
print("图1生成完毕！")
```

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['font.sans-serif'] = ['SimHei'] 
plt.rcParams['axes.unicode_minus'] = False   
x = np.linspace(-3, 3, 200)
y = np.linspace(-3, 3, 200)
X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2) + 1e-5
Z = 0.5 * R**2 - np.log(R) 
plt.figure(figsize=(8, 6))
contour = plt.contourf(X, Y, Z, levels=30, cmap='viridis')
plt.colorbar(contour, label='Time Delay (arbitrary units)')
plt.scatter(0.5, 0, color='red', s=100, marker='*', label='Image 1 (极值点)')
plt.scatter(-0.5, 0, color='orange', s=100, marker='*', label='Image 2 (极值点)')
plt.title('Fermat Potential / Time Delay Distribution', fontsize=14)
plt.xlabel('Image Plane X (arcsec)', fontsize=12)
plt.ylabel('Image Plane Y (arcsec)', fontsize=12)
plt.legend()
plt.tight_layout()
plt.savefig('Figure2_TimeDelayMap.png', dpi=300)
plt.show()
print("图2（费马势热力图）生成完毕！")
import matplotlib.pyplot as plt
```

```python
import matplotlib.pyplot as plt
import numpy as np
phase = np.linspace(0, 2*np.pi, 100) 
delta_t = 2e-6 * np.sin(phase) 
plt.figure(figsize=(8, 5))
plt.plot(phase, delta_t, color='purple', linewidth=2, label='Residual Modulation')
plt.axhline(0, color='gray', linestyle='--') 
plt.title('Modulation of Time Delay Due to Earth\'s Orbital Motion', fontsize=14)
plt.xlabel('Earth Orbital Phase (rad)', fontsize=12)
plt.ylabel('Residual Time Delay (days)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.savefig('Figure3_ObserverMotion.png', dpi=300)
plt.show()
print("图3（观测者运动调制图）生成完毕！")
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(-2, 2, 15)
y = np.linspace(-2, 2, 15)
X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2) + 1e-5 
U = -Y / R
V = X / R
strength = 1.0 / R
U = U * strength
V = V * strength
plt.figure(figsize=(8, 8))
plt.quiver(X, Y, U, V, color='green', pivot='mid', scale=20, width=0.005)
plt.scatter(0, 0, color='red', s=200, label='Lens Galaxy (前景)', zorder=5)
plt.title('Weak Lensing Shear Field (Vector Representation)', fontsize=14)
plt.xlabel('Image Plane X (arcsec)', fontsize=12)
plt.ylabel('Image Plane Y (arcsec)', fontsize=12)
plt.xlim(-2.5, 2.5)
plt.ylim(-2.5, 2.5)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right')
plt.tight_layout()
plt.savefig('Figure4_ShearField.png', dpi=300)
plt.show()
print("图4（弱透镜剪切场矢量图）生成完毕！")
```
